```python
import importlib
import unittest
import networkx as nx # In case your tests need it

# Load the module
nb_module = importlib.import_module("coding1-py-vaccaro")

# Extract functions into the local namespace so you can call them directly
substring_shared = nb_module.substring_shared
find_kmer_hashing = nb_module.find_kmer_hashing
find_rc_palindromes_naive = nb_module.find_rc_palindromes_naive
find_rc_palindrome_starts = nb_module.find_rc_palindrome_starts
build_adjlist = nb_module.build_adjlist
bfs = nb_module.bfs
dijkstra = nb_module.dijkstra
bellman_ford = nb_module.bellman_ford
load_network = nb_module.load_network
_reconstruct = nb_module._reconstruct

print("Functions successfully imported!")
```

```python
class TestSubstringAlgorithms(unittest.TestCase):
    
    def test_shared_substring_naive(self):
        self.assertEqual(substring_shared("ACGT", "CG", 4, 2, 2), 1)
        self.assertEqual(substring_shared("ACGT", "ZZ", 4, 2, 2), 0)

    def test_shared_substring_hashing(self):
        self.assertEqual(find_kmer_hashing("ACGT", "CG", 4, 2, 2), "CG")
        self.assertIsNone(find_kmer_hashing("ACGT", "ZZ", 4, 2, 2))

    def test_edge_cases(self):
        self.assertEqual(find_kmer_hashing("ACGT", "CG", 4, 2, 0), "Invalid k value, insert a k of at least two")
        self.assertIsNone(find_kmer_hashing("ACGT", "CG", 4, 2, 5))

suite1 = unittest.TestLoader().loadTestsFromTestCase(TestSubstringAlgorithms)
unittest.TextTestRunner(verbosity=2).run(suite1)
```

```python
class TestPalindromeAlgorithms(unittest.TestCase):
    
    def test_rc_palindromes_naive(self):
        # "AACGAATTCGTT" -> GAATTC starts at index 3
        seq = "AACGAATTCGTT"
        starts = find_rc_palindromes_naive(seq, k=6)
        self.assertIn(3, starts)

    def test_rc_palindromes_expansion(self):
        seq = "AACGAATTCGTT"
        starts = find_rc_palindrome_starts(seq, k=6)
        self.assertIn(3, starts)

    def test_consistency(self):
        seq = "GAATTCATCGATGCGC"
        k = 4
        self.assertEqual(
            sorted(find_rc_palindromes_naive(seq, k)),
            sorted(find_rc_palindrome_starts(seq, k))
        )

suite2 = unittest.TestLoader().loadTestsFromTestCase(TestPalindromeAlgorithms)
unittest.TextTestRunner(verbosity=2).run(suite2)
```

```python
class TestGraphAlgorithms(unittest.TestCase):
    
    def setUp(self):
        self.G = nx.DiGraph()
        self.G.add_edge("A", "B", weight=1.0)
        self.G.add_edge("B", "C", weight=2.0)
        self.G.add_edge("A", "C", weight=5.0)
        self.adj = build_adjlist(self.G)

    def test_bfs(self):
        dist, paths = bfs(self.adj, "A")
        self.assertEqual(dist["C"], 1)  # Direct edge A -> C is 1 hop
        self.assertEqual(paths["C"], ["A", "C"])

    def test_dijkstra(self):
        dist, paths = dijkstra(self.adj, "A")
        self.assertEqual(dist["C"], 3.0)  # Path A -> B -> C (1.0 + 2.0 = 3.0)
        
        # Check parent node directly
        self.assertEqual(paths["C"], "B")

    def test_bellman_ford_negative_cycle(self):
        neg_graph = {"A": [("B", 1.0)], "B": [("A", -2.0)]}
        nodes = {"A", "B"}
        _, _, neg_cycle = bellman_ford(neg_graph, nodes, "A")
        self.assertIsNotNone(neg_cycle)

suite3 = unittest.TestLoader().loadTestsFromTestCase(TestGraphAlgorithms)
unittest.TextTestRunner(verbosity=2).run(suite3)
```
